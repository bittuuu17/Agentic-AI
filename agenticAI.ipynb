````python
class SimpleAgent:
    def __init__(self):
        self.memory = []  # store last 3 percepts

    def perceive(self, input_data):
        self.memory.append(input_data)
        if len(self.memory) > 3:
            self.memory.pop(0)

    def decide(self):
        if "goal" in self.memory:
            return "take_action"
        return "idle"

def main():
    agent = SimpleAgent()
    percepts = ["noise", "goal", "distraction", "goal"]
    for p in percepts:
        agent.perceive(p)
        print(f"Percepts: {agent.memory} → Action: {agent.decide()}")

# Call main
if __name__ == "__main__":
    main()
````

````python
class HelloAgent:
    def __init__(self):
        self.memory = []
        self.optional = ["Hello", "World"]

    def perceive(self, input_data):
        self.memory.append(input_data)

    def reactCaseSensitive(self):
        if "hello" in self.memory:
            print("Hello")
        if "world" in self.memory:
            print("World")
        if  any(item in self.memory for item in self.optional):
            print("Hello World")

    def reactNotCaseSensitive(self):
        optional = {item.lower() for item in self.optional}
        if all(item.lower() in self.memory for item in optional):
            print("Hello World")

agent = HelloAgent()
agent.perceive("hello")
agent.perceive("world")
agent.reactCaseSensitive()
agent.reactNotCaseSensitive()




````

# Day 5 – LLM-Powered Planning Agent (Perplexity)

## Overview
Day 5 marks the transition from rule-based agents to a **true agentic AI system**.  
In this stage, the agent no longer relies on hardcoded logic to decide actions. Instead, it uses a **Large Language Model (Perplexity – sonar-pro)** as a planner to dynamically select tools based on the current goal and internal state.

This architecture follows the principle:

> **LLMs reason, tools execute, state remembers**

---

## Objective
Build an agent that:
- Uses an LLM to decide which tool to invoke
- Maintains internal memory (state)
- Executes deterministic tools
- Separates reasoning from execution
- Aligns with modern agent frameworks and MCP concepts

---

## Agent Architecture




````python
import requests
from google.colab import userdata


API_KEY = userdata.get("PERPLEXITY_API_KEY")

def perplexity_decided_tool(goal, state):
    # Perplexity requires "Bearer <token>" format
    headers = {
        "Authorization": f"Bearer {API_KEY}",
        "Content-Type": "application/json",
        "accept": "application/json"
    }

    # Perplexity uses the Chat Completions format
    messages = [
        {"role": "system", "content": "You are a specialized tool-decision agent. Return only the tool name."},
        {"role": "user", "content": f"Goal: {goal}\nCurrent state: {state}\nDecide tool: get_weather / send_message / none."}
    ]

    data = {
        "model": "sonar-pro", # Valid 2025 model
        "messages": messages,
        "temperature": 0  # Set to 0 for consistent tool selection
    }

    # Use the correct API endpoint
    response = requests.post(
        "https://api.perplexity.ai/chat/completions",
        headers=headers,
        json=data
    )

    if response.status_code == 200:
        result = response.json()
        # Extract response text from the chat completion object
        return result["choices"][0]["message"]["content"].strip().lower()
    else:
        print(f"Error {response.status_code}: {response.text}")
        return "none"



def get_weather(city):
  weather_date={"New York" : "rainy", "Mumbai" : "sunny"}
  return weather_date.get(city, "not determined")

def send_message(msg):
    print(f"Sending Message: {msg}")

class LLMPlanningAgent:

    def __init__(self, city, goal):
        self.city = city
        self.goal = goal
        self.state = {}

    def decide_tool(self):
        print("deciding tool")
        return perplexity_decided_tool(self.goal, self.state)

    def act(self):
        print(f"Planning to {self.goal}")
        tool = self.decide_tool()
        print(f"Using tool choosen by perplexity: {tool}")

        if tool == "get_weather":
            self.state["weather"]= get_weather(self.city)
            print(f"Weather in {self.city} is {self.state['weather']}")
        elif tool == "send_message":
            if "weather" in self.state:
                 message = f"Weather in {self.city} is {self.state['weather']}"
                 send_message(message)
        else:
            print("no action")


def main():
    agent = LLMPlanningAgent(goal="Check weather", city="New York")
    agent.act()

if __name__ == "__main__":
    main()
````

## Day 8 – Multi-Goal Priority Agent

### New Concepts
- Goal queues
- Priority-based planning
- Dynamic goal selection
- State-driven completion

### Architecture
Agent selects the highest-priority unfinished goal at each step and uses
an LLM to decide actions based on current state.

### Insight
This pattern enables autonomous systems to handle complex objectives
without hardcoded execution order.


````python
import requests
from google.colab import userdata

API_KEY = userdata.get("PERPLEXITY_API_KEY")

def get_weather(city):
    weather_data={"New York": "rainy", "Mumbai": "sunny"}
    return weather_data.get(city, "not determined")

def send_message(msg):
    print(f"Sending Message: {msg}")

def perplexity_decided_tool(goal, state):

  headers = {
      "Authorization" : f"Bearer {API_KEY}",
      "Content_Type": "application/json",
      "accept": "application/json"
  }

  messages = [
      {
          "role" : "system",
          "content": (
              "You are a planning agent"
              "Return only the tool name. based on goal once done send extra advice with done reply"
              "get_weather, send_message, done"
          )
      },
      {
          "role": "user",
          "content": f"Goal: {goal}\nState:{state}"
      }
  ]

  data={
      "model": "sonar-pro",
      "messages": messages,
      "temperature": 0
  }

  response = requests.post(
      "https://api.perplexity.ai/chat/completions",
      headers=headers,
      json=data
  )
  print(response.json())
  return response.json()["choices"][0]["message"]["content"].strip().lower()

class Agent:
  def __init__(self, city):
      self.city=city;
      self.goal=[
            {"name": "get weather", "priority": 1, "done": False},
            {"name": "send advice", "priority": 2, "done": False},
      ]
      self.state={
          "weather": None,
          "message_sent": False,
          "steps":[]
      }

  def get_active_goal(self):
      return min(
            (g for g in self.goal if not g["done"]),
        key= lambda g: g["priority"], default=None)


  def act(self):
      print("Agent Decision Loop Starts")

      while True:
          active_goal= self.get_active_goal()
          print(f"Active Goal: {active_goal}")

          if active_goal is None:
                  print("Goal Achieved")
                  break

          tool= perplexity_decided_tool(active_goal["name"], self.state)
          print(f"Choosen Tool: {tool}")
          self.state["steps"].append(tool)

          if tool == "get_weather":
              self.state["weather"]= get_weather(self.city)
              print(f"Weather in {self.city} is {self.state['weather']}")
              active_goal["done"]=True

          if tool == "send_message":
              if "weather" in self.state and not self.state["message_sent"]:
                  send_message(f"Weather in {self.city} is {self.state['weather']}")
                  self.state["message_sent"] = True
                  active_goal["done"]=True

          if tool =="done":
              print("Goal Achieved")
              break
          else:
              print("No action")
              break

def main():
    agent = Agent(city="New York")
    agent.act()


if __name__ == "__main__":
    main()
````

# Day 9: Building Robust Agents

## Key Improvements Made:

### 1. Error Handling with Retries
**Problem:** API calls can fail (network, rate limits, timeouts)
**Solution:** Retry logic with exponential backoff

### 2. Structured Outputs (JSON)
**Problem:** LLM can return freeform text, hard to parse
**Solution:** Request JSON format: `{"tool": "name", "reason": "why"}`

### 3. Precondition Checking
**Problem:** Agent tries to execute actions without prerequisites
**Solution:** Check state before execution (e.g., can't send message without weather)

### 4. Iteration Limits
**Problem:** Infinite loops if agent gets stuck
**Solution:** `max_iterations` parameter (typically 10-20)

### 5. Comprehensive Logging
**Problem:** Hard to debug agent decisions
**Solution:** Log every step, state changes, LLM responses

````python
import requests
from google.colab import userdata
import json
import time

API_KEY = userdata.get("PERPLEXITY_API_KEY")

# ========================================
# HELPER FUNCTIONS
# ========================================

def clean_json_response(text):
    """Remove markdown code fences from LLM response"""
    text = text.strip()

    if text.startswith("```json"):
        text = text[7:]
    elif text.startswith("```python"):
        text = text[9:]
    elif text.startswith("```"):
        text = text[3:]

    if text.endswith("```"):
        text = text[:-3]

    return text.strip()

# ========================================
# TOOL FUNCTIONS
# ========================================

def get_weather(city):
    """Simulate weather API"""
    import random
    weather_data = {
        "New York": ["rainy", "cloudy", "sunny"],
        "Mumbai": ["sunny", "humid", "cloudy"],
        "Tokyo": ["rainy", "foggy", "snowy"],
        "London": ["sunny", "cloudy", "rainy"]
    }
    if city in weather_data:
        return random.choice(weather_data[city])
    else:
        return "unknown"

def send_message(msg):
    """Send message to user"""
    print("Sending Message:")
    print(msg)
    time.sleep(0.3)
    print("Message Sent Successfully")

# ========================================
# LLM TOOL SELECTOR & ADVISOR
# ========================================

def perplexity_decided_tool_Advisor(goal, state, current_weather=None, city=None, max_retries=3):
    """
    Dual-purpose function:
    1. If current_weather is None → Select tool
    2. If current_weather is provided → Generate advice
    """

    headers = {
        "Authorization": f"Bearer {API_KEY}",
        "Content-Type": "application/json",
        "accept": "application/json"
    }

    # MODE 1: TOOL SELECTION
    if current_weather is None:
        system_prompt = """You are a tool selector for an AI agent.

Your job: Select EXACTLY ONE tool to execute next based on the current state.

Available tools:
- get_weather
- send_message
- done

Rules:
1. Return ONLY valid JSON: {"tool": "tool_name", "reason": "why"}
2. tool_name must be one of: get_weather, send_message, done
3. If weather is null → get_weather
4. If weather exists and message_sent is false → send_message
5. If weather exists and message_sent is true → done
"""

        messages = [
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": f"Goal: {goal}\nState:\n{json.dumps(state, indent=2)}"}
        ]
        temperature = 0

    # MODE 2: ADVICE GENERATION
    else:
        system_prompt_advice = """You are a weather advisor.

Return ONLY JSON: {"advice": "one sentence advice"}.
"""

        messages = [
            {"role": "system", "content": system_prompt_advice},
            {"role": "user", "content": f"City: {city}\nWeather: {current_weather}"}
        ]
        temperature = 1

    data = {
        "model": "sonar-pro",
        "messages": messages,
        "temperature": temperature
    }

    for attempt in range(max_retries):
        try:
            mode = "TOOL SELECTION" if current_weather is None else "ADVICE GENERATION"
            print(f"LLM Request [{mode}] Attempt {attempt + 1}")

            response = requests.post(
                "https://api.perplexity.ai/chat/completions",
                json=data,
                headers=headers,
                timeout=30
            )
            response.raise_for_status()

            result = response.json()["choices"][0]["message"]["content"]
            cleaned = clean_json_response(result)

            parsed = json.loads(cleaned)

            if current_weather is None:
                return parsed.get("tool", "get_weather")
            else:
                return json.dumps(parsed)

        except Exception:
            if attempt < max_retries - 1:
                time.sleep(2 ** attempt)

    if current_weather is None:
        return "get_weather"
    else:
        return json.dumps({"advice": "Have a great day and stay safe!"})

# ========================================
# AGENT CLASS
# ========================================

class Agent:
    def __init__(self, city):
        self.city = city
        self.goal = [
            {"name": "get weather", "priority": 1, "done": False},
            {"name": "send advice", "priority": 2, "done": False},
        ]
        self.state = {
            "city": city,
            "weather": None,
            "message_sent": False,
            "steps": []
        }
        self.max_iterations = 10

    def get_active_goal(self):
        incomplete = [g for g in self.goal if not g["done"]]
        if not incomplete:
            return None
        return min(incomplete, key=lambda g: g["priority"])

    def execute_tool(self, tool, active_goal):

        if tool == "get_weather":
            print("Executing get_weather")
            weather = get_weather(self.city)
            self.state["weather"] = weather
            active_goal["done"] = True
            return True

        elif tool == "send_message":
            advice_response = perplexity_decided_tool_Advisor(
                goal="provide weather advice",
                state=self.state,
                current_weather=self.state["weather"],
                city=self.city
            )

            advice = json.loads(advice_response).get("advice", "")

            msg = f"""Weather Report for {self.city}
--------------------------------
Current Weather: {self.state['weather']}
Advice: {advice}
--------------------------------
"""
            send_message(msg)

            self.state["message_sent"] = True
            active_goal["done"] = True
            return True

        elif tool == "done":
            return True

        return False

    def act(self):
        for iteration in range(self.max_iterations):
            active_goal = self.get_active_goal()
            if not active_goal:
                break

            tool = perplexity_decided_tool_Advisor(
                goal=active_goal["name"],
                state=self.state
            )

            self.state["steps"].append({
                "iteration": iteration + 1,
                "goal": active_goal["name"],
                "tool": tool
            })

            self.execute_tool(tool, active_goal)

# ========================================
# MAIN
# ========================================

def main():
    agent = Agent(city="New York")
    agent.act()

if __name__ == "__main__":
    main()
````

````python
"""Day 10 - Multi-Tool Agent with Real APIs (All Options)

APIS CONFIGURED:
Perplexity - LLM reasoning (you have)
yfinance - Stock data (no key needed)
Google News RSS - News (no key needed)
wttr.in - Weather fallback (no key needed)
Exchange Rate API - Currency (no key needed)

NewsAPI - news (sign up: https://newsapi.org)
WeatherAPI - weather (sign up: https://weatherapi.com)
OpenWeather - Weather alternative (sign up: https://openweathermap.org)
Alpha Vantage - Financial data (sign up: https://alphavantage.co)
"""

import requests
from google.colab import userdata
import json
import time

!pip install yfinance beautifulsoup4 lxml -q

import yfinance as yf
from bs4 import BeautifulSoup
from datetime import datetime

try:
  PERPLEXITY_API_KEY = userdata.get("PERPLEXITY_API_KEY")
except:
  PERPLEXITY_API_KEY = None
  print(f"Perplexity API Key not found")


try:
    NEWSAPI_KEY = userdata.get("NEWSAPI_KEY")
    print("NewsAPI key loaded")
except:
    NEWSAPI_KEY = None
    print("NewsAPI key not found - using Google News RSS fallback")

# WeatherAPI.com (Best weather option - instant activation!)
try:
    WEATHERAPI_KEY = userdata.get("WEATHERAPI_KEY")
    print("WeatherAPI key loaded")
except:
    WEATHERAPI_KEY = None
    print("WeatherAPI key not found - using wttr.in fallback")

# OpenWeather (Alternative weather - 10 min activation wait)
try:
    OPENWEATHER_KEY = userdata.get("OPENWEATHER_API_KEY")
    print("OpenWeather key loaded")
except:
    OPENWEATHER_KEY = None
    print("OpenWeather key not found - using wttr.in fallback")

# Alpha Vantage (Financial data)
try:
    ALPHA_VANTAGE_KEY = userdata.get("ALPHA_VANTAGE_KEY")
    print("Alpha Vantage key loaded")
except:
    ALPHA_VANTAGE_KEY = None
    print("Alpha Vantage key not found - using yfinance only")

# ========================================
# HELPER FUNCTIONS
# ========================================

def clean_json_response(response_data):
    text=response_data.strip()

    if text.startswith("```json"):
        text = text[7:]
    elif text.startswith("```python"):
        text = text[9:]
    elif text.startswith("```"):
        text = text[3:]

    if text.endswith("```"):
        text = text[:-3]

    return text.strip()

def get_stock_price(ticker):
    """
    Fetch real-time stock price using Yahoo Finance API.
    Handles weekends, holidays, and missing data gracefully.

    Args:
        ticker (str): Stock symbol (e.g., "AAPL", "GOOGL", "TSLA")

    Returns:
        dict: Stock data with price, change, volume, and metadata
    """
    print(f"Fetching stock data for {ticker}...")

    try:
        stock = yf.Ticker(ticker)
        info = stock.info
        history = stock.history(period="5d")

        if history.empty:
            print(f"No recent data for {ticker}")
            return {"error": f"No data available for {ticker}"}

        current_price = history['Close'].iloc[-1]

        if len(history) > 1:
            previous_close = history['Close'].iloc[-2]
            change_percent = ((current_price - previous_close) / previous_close) * 100
        else:
            change_percent = 0.0

        volume = history['Volume'].iloc[-1] if 'Volume' in history.columns else 0

        data = {
            "ticker": ticker.upper(),
            "price": round(float(current_price), 2),
            "change_percent": round(float(change_percent), 2),
            "volume": f"{int(volume):,}" if volume > 0 else "N/A",
            "market_cap": f"${info.get('marketCap', 0) / 1e9:.2f}B" if info.get('marketCap') else "N/A",
            "pe_ratio": round(float(info.get('trailingPE', 0)), 2) if info.get('trailingPE') else "N/A",
            "company_name": info.get('longName') or info.get('shortName') or ticker.upper(),
            "last_updated": str(history.index[-1].date()),
            "source": "Yahoo Finance"
        }

        print(f"Retrieved: {data['company_name']}: ${data['price']} ({data['change_percent']:+.2f}%)")
        print(f"Data as of: {data['last_updated']}")

        from datetime import datetime
        today = datetime.now().date()
        data_date = history.index[-1].date()
        days_old = (today - data_date).days

        if days_old > 0:
            print(f"Note: Markets closed, showing last trading day ({days_old} day(s) old)")

        return data

    except Exception as e:
        print(f"Error: {e}")


def get_news(query, max_results=3):
    """
    Get REAL news headlines with automatic fallback

    Priority:
    1. NewsAPI (if key available) - Best quality
    2. Google News RSS (always free) - Good quality

    Args:
        query (str): Search query (e.g., "AAPL", "Tesla", "AI")
        max_results (int): Number of headlines to return (default: 3)

    Returns:
        list: News headlines with title, source, date, url
    """
    print(f"  📰 Fetching REAL news for '{query}'...")

    # METHOD 1: NewsAPI (if key available)
    if NEWSAPI_KEY:
        try:
            url = "https://newsapi.org/v2/everything"
            params = {
                "q": query,
                "apiKey": NEWSAPI_KEY,
                "pageSize": max_results,
                "sortBy": "publishedAt",
                "language": "en"
            }

            response = requests.get(url, params=params, timeout=10)

            if response.status_code == 200:
                data = response.json()
                articles = data.get('articles', [])

                headlines = []
                for article in articles[:max_results]:
                    headlines.append({
                        "title": article['title'],
                        "source": article['source']['name'],
                        "published": article['publishedAt'][:10],
                        "url": article['url']
                    })

                print(f"  ✅ Found {len(headlines)} news articles (NewsAPI)")
                return headlines
            else:
                print(f"NewsAPI error {response.status_code}, trying Google News...")
        except Exception as e:
            print(f"NewsAPI failed: {e}, trying Google News...")

    # METHOD 2: Google News RSS (always free, no key needed!)
    try:
        # Google News RSS feed - completely free!
        rss_url = f"https://news.google.com/rss/search?q={query}&hl=en-US&gl=US&ceid=US:en"

        response = requests.get(rss_url, timeout=10)
        soup = BeautifulSoup(response.content, 'xml')

        items = soup.find_all('item')[:max_results]

        headlines = []
        for item in items:
            headlines.append({
                "title": item.title.text if item.title else "No title",
                "source": item.source.text if item.source else "Google News",
                "published": item.pubDate.text[:16] if item.pubDate else "Recent",
                "url": item.link.text if item.link else ""
            })

        print(f"Found {len(headlines)} news articles (Google News RSS)")
        return headlines

    except Exception as e:
        print(f"Error fetching news: {e}")
        return [{"title": "News unavailable", "source": "Error", "published": "", "url": ""}]

def get_weather(city):
    """
    Fetch REAL weather data with automatic fallback

    Priority:
    1. WeatherAPI.com (if key available) - Best, instant activation
    2. OpenWeather (if key available) - Good, 10-min wait after signup
    3. wttr.in (always free) - Good fallback, no key needed

    Args:
        city (str): City name (e.g., "New York", "London", "Tokyo")

    Returns:
        dict: Weather information (condition, temp, humidity, wind)
    """
    print(f"Fetching REAL weather for {city}...")

    # METHOD 1: WeatherAPI.com (if key available - BEST OPTION)
    if WEATHERAPI_KEY:
        try:
            url = "http://api.weatherapi.com/v1/current.json"
            params = {
                "key": WEATHERAPI_KEY,
                "q": city,
                "aqi": "no"  # Don't need air quality data
            }

            response = requests.get(url, params=params, timeout=10)

            if response.status_code == 200:
                data = response.json()

                weather = {
                    "city": data['location']['name'],
                    "country": data['location']['country'],
                    "condition": data['current']['condition']['text'],
                    "temperature": f"{data['current']['temp_c']}°C",
                    "feels_like": f"{data['current']['feelslike_c']}°C",
                    "humidity": f"{data['current']['humidity']}%",
                    "wind_speed": f"{data['current']['wind_kph']} km/h",
                    "last_updated": data['current']['last_updated'],
                    "source": "WeatherAPI.com"
                }

                print(f"{weather['city']}: {weather['condition']}, {weather['temperature']} (WeatherAPI)")
                return weather
            else:
                print(f"WeatherAPI error {response.status_code}, trying fallback...")
        except Exception as e:
            print(f"WeatherAPI failed: {e}, trying fallback...")

    # METHOD 2: OpenWeather (if key available and activated)
    if OPENWEATHER_KEY:
        try:
            url = "http://api.openweathermap.org/data/2.5/weather"
            params = {
                "q": city,
                "appid": OPENWEATHER_KEY,
                "units": "metric"  # Celsius
            }

            response = requests.get(url, params=params, timeout=10)

            if response.status_code == 200:
                data = response.json()

                weather = {
                    "city": data['name'],
                    "country": data['sys']['country'],
                    "condition": data['weather'][0]['description'],
                    "temperature": f"{data['main']['temp']}°C",
                    "feels_like": f"{data['main']['feels_like']}°C",
                    "humidity": f"{data['main']['humidity']}%",
                    "wind_speed": f"{data['wind']['speed']} m/s",
                    "source": "OpenWeather"
                }

                print(f"{weather['city']}: {weather['condition']}, {weather['temperature']} (OpenWeather)")
                return weather
            elif response.status_code == 401:
                print(f"OpenWeather key not activated yet (wait 10-15 mins), trying wttr.in...")
            else:
                print(f"OpenWeather error {response.status_code}, trying wttr.in...")
        except Exception as e:
            print(f"OpenWeather failed: {e}, trying wttr.in...")

    # METHOD 3: wttr.in (always free, no key needed - RELIABLE FALLBACK)
    try:
        # wttr.in API - completely free and unlimited!
        url = f"https://wttr.in/{city}?format=j1"

        response = requests.get(url, timeout=10)

        if response.status_code == 200:
            data = response.json()
            current = data['current_condition'][0]

            weather = {
                "city": city,
                "condition": current['weatherDesc'][0]['value'],
                "temperature": f"{current['temp_C']}°C",
                "feels_like": f"{current['FeelsLikeC']}°C",
                "humidity": f"{current['humidity']}%",
                "wind_speed": f"{current['windspeedKmph']} km/h",
                "source": "wttr.in"
            }

            print(f"{weather['city']}: {weather['condition']}, {weather['temperature']} (wttr.in)")
            return weather
        else:
            return {"error": "Weather data unavailable"}

    except Exception as e:
        print(f"Error: {e}")
        return {"error": str(e)}

def calculator(expression):
    """
    Perform mathematical calculations (completely free!)

    Args:
        expression (str): Math expression (e.g., "178.50 * 100", "25 + 30")

    Returns:
        float: Calculation result
    """
    print(f"Calculating: {expression}")

    try:
        # Safe evaluation - only allow numbers and basic operators
        allowed_chars = "0123456789+-*/(). "

        if all(c in allowed_chars for c in expression):
            result = eval(expression)
            print(f"Result: {result}")
            return result
        else:
            print(f"Expression contains invalid characters")
            return None

    except Exception as e:
        print(f"Calculation error: {e}")
        return None

def web_search(query):
    """
    Search the web using Perplexity API (you already have this!)

    Args:
        query (str): Search query

    Returns:
        str: Search result summary
    """
    print(f"Searching web for '{query}'...")

    try:
        headers = {
            "Authorization": f"Bearer {PERPLEXITY_API_KEY}",
            "Content-Type": "application/json"
        }

        data = {
            "model": "sonar-pro",
            "messages": [
                {
                    "role": "system",
                    "content": "You are a helpful assistant. Provide concise, factual answers in 2-3 sentences."
                },
                {
                    "role": "user",
                    "content": query
                }
            ],
            "temperature": 0.2
        }

        response = requests.post(
            "https://api.perplexity.ai/chat/completions",
            headers=headers,
            json=data,
            timeout=20
        )

        result = response.json()["choices"][0]["message"]["content"]
        print(f"Search completed")
        return result

    except Exception as e:
        print(f"Search failed: {e}")
        return f"Search unavailable: {str(e)}"

def get_company_info(ticker):
    """
    Get company information (FREE, no API key!)

    Uses yfinance for comprehensive company data

    Args:
        ticker (str): Stock ticker (e.g., "AAPL", "GOOGL")

    Returns:
        dict: Company information (name, sector, industry, employees, etc.)
    """
    print(f"Fetching company info for {ticker}...")

    try:
        # Use yfinance - free and comprehensive
        stock = yf.Ticker(ticker)
        info = stock.info

        company_data = {
            "ticker": ticker.upper(),
            "name": info.get('longName', ticker),
            "sector": info.get('sector', 'N/A'),
            "industry": info.get('industry', 'N/A'),
            "employees": f"{info.get('fullTimeEmployees', 0):,}" if info.get('fullTimeEmployees') else 'N/A',
            "description": info.get('longBusinessSummary', 'N/A')[:200] + "..." if info.get('longBusinessSummary') else 'N/A',
            "website": info.get('website', 'N/A'),
            "headquarters": f"{info.get('city', '')}, {info.get('country', '')}".strip(', '),
            "source": "Yahoo Finance"
        }

        print(f"Found: {company_data['name']}")
        return company_data

    except Exception as e:
        print(f"Error: {e}")
        return {"error": str(e)}

def get_exchange_rate(from_currency="USD", to_currency="EUR"):
    """
    Get currency exchange rates (FREE, no API key needed!)

    Uses exchangerate-api.com (1500 requests/month free)

    Args:
        from_currency (str): Source currency (e.g., "USD", "EUR", "GBP")
        to_currency (str): Target currency (e.g., "EUR", "JPY", "INR")

    Returns:
        dict: Exchange rate information
    """
    print(f"Fetching exchange rate {from_currency} → {to_currency}...")

    try:
        url = f"https://api.exchangerate-api.com/v4/latest/{from_currency}"

        response = requests.get(url, timeout=10)

        if response.status_code == 200:
            data = response.json()
            rate = data['rates'].get(to_currency)

            if rate:
                result = {
                    "from": from_currency,
                    "to": to_currency,
                    "rate": round(rate, 4),
                    "date": data['date'],
                    "source": "ExchangeRate-API"
                }
                print(f"1 {from_currency} = {rate:.4f} {to_currency}")
                return result
            else:
                return {"error": f"Currency {to_currency} not found"}
        else:
            return {"error": "Exchange rate API unavailable"}

    except Exception as e:
        print(f"Error: {e}")
        return {"error": str(e)}

def send_message(message):
    """
    Send message to user (free, always works!)

    Args:
        message (str): Message content

    Returns:
        bool: Success status
    """
    print(f"\n{'='*60}")
    print(f"MESSAGE TO USER:")
    print(f"{'='*60}")
    print(message)
    print(f"{'='*60}\n")
    time.sleep(0.3)
    return True

class ToolRegistry:
    """
    Centralized registry for all available tools
    Manages tool execution and provides descriptions to LLM
    """

    def __init__(self):
        self.tools = {
            "get_stock_price": {
                "function": get_stock_price,
                "description": "Fetch REAL stock price and data from Yahoo Finance",
                "parameters": ["ticker (str): Stock symbol like AAPL, GOOGL, TSLA, MSFT"],
                "example": "get_stock_price('AAPL')",
                "free": True,
                "requires_key": False
            },
            "get_news": {
                "function": get_news,
                "description": "Get REAL news headlines from NewsAPI or Google News",
                "parameters": ["query (str): Search term", "max_results (int): Number of results (default 3)"],
                "example": "get_news('Tesla', 5)",
                "free": True,
                "requires_key": False  # Has free fallback
            },
            "get_weather": {
                "function": get_weather,
                "description": "Fetch REAL weather data from WeatherAPI, OpenWeather, or wttr.in",
                "parameters": ["city (str): City name like 'New York', 'London', 'Tokyo'"],
                "example": "get_weather('New York')",
                "free": True,
                "requires_key": False  # Has free fallback
            },
            "calculator": {
                "function": calculator,
                "description": "Perform mathematical calculations",
                "parameters": ["expression (str): Math expression like '178.50 * 100'"],
                "example": "calculator('178.50 * 100')",
                "free": True,
                "requires_key": False
            },
            "web_search": {
                "function": web_search,
                "description": "Search the web using Perplexity AI for current information",
                "parameters": ["query (str): Search query"],
                "example": "web_search('What is P/E ratio?')",
                "free": False,  # Uses your Perplexity credits
                "requires_key": True
            },
            "get_company_info": {
                "function": get_company_info,
                "description": "Get company information from Yahoo Finance (sector, industry, employees, etc.)",
                "parameters": ["ticker (str): Stock ticker like AAPL, GOOGL"],
                "example": "get_company_info('AAPL')",
                "free": True,
                "requires_key": False
            },
            "get_exchange_rate": {
                "function": get_exchange_rate,
                "description": "Get currency exchange rates between two currencies",
                "parameters": ["from_currency (str): Source currency", "to_currency (str): Target currency"],
                "example": "get_exchange_rate('USD', 'EUR')",
                "free": True,
                "requires_key": False
            },
            "send_message": {
                "function": send_message,
                "description": "Send final message/report to user",
                "parameters": ["message (str): Message content"],
                "example": "send_message('Analysis complete')",
                "free": True,
                "requires_key": False
            },
            "done": {
                "function": None,
                "description": "Mark task as complete (no more actions needed)",
                "parameters": [],
                "example": "done",
                "free": True,
                "requires_key": False
            }
        }

    def get_tool_list_for_llm(self):
        """Generate formatted tool list for LLM prompt"""
        tool_descriptions = []

        for name, info in self.tools.items():
            if name != "done":
                # Show if free and if key is needed
                status = "FREE" if info.get("free", True) else "Uses credits"
                if info.get("requires_key") and name != "web_search":
                    status += " (has fallback)"

                desc = f"- {name}: {info['description']} ({status})"
                tool_descriptions.append(desc)

        tool_descriptions.append("- done: Mark task as complete")
        return "\n".join(tool_descriptions)

    def execute(self, tool_name, **kwargs):
        """
        Execute a tool by name with parameters

        Args:
            tool_name (str): Name of tool to execute
            **kwargs: Tool-specific parameters

        Returns:
            Any: Tool execution result
        """
        if tool_name not in self.tools:
            print(f"Unknown tool: {tool_name}")
            return None

        if tool_name == "done":
            return "Task marked as complete"

        tool_func = self.tools[tool_name]["function"]

        try:
            # Execute tool with provided parameters
            result = tool_func(**kwargs)
            return result
        except Exception as e:
            print(f"Tool execution error: {e}")
            return None

# ========================================
# LLM DECISION MAKER
# ========================================

def ask_llm_for_next_action(task, state, tool_registry, max_retries=3):
    """
    Ask LLM to decide next action (tool + parameters)

    Args:
        task (str): Original user task
        state (dict): Current agent state
        tool_registry (ToolRegistry): Available tools
        max_retries (int): Retry attempts

    Returns:
        dict: {"tool": "tool_name", "params": {...}, "reason": "why"}
    """

    headers = {
        "Authorization": f"Bearer {PERPLEXITY_API_KEY}",
        "Content-Type": "application/json"
    }

    tool_list = tool_registry.get_tool_list_for_llm()

    system_prompt = f"""You are an intelligent AI agent that selects tools to complete tasks.

Available tools (ALL WITH REAL DATA!):
{tool_list}

Your job:
1. Analyze the current state and task
2. Decide which tool to use next
3. Provide parameters for the tool
4. Return ONLY valid JSON (no markdown, no code fences)

Response format:
{{
    "tool": "tool_name",
    "params": {{"param1": "value1"}},
    "reason": "why this tool is needed now"
}}

Rules:
- All tools use REAL APIs (stock prices, news, weather are live!)
- Use tools in logical order (e.g., get data before analyzing)
- Don't repeat tools unless necessary
- Use 'done' when task is complete
- Be efficient (don't use unnecessary tools)

Example responses:
{{"tool": "get_stock_price", "params": {{"ticker": "AAPL"}}, "reason": "Need current price for analysis"}}
{{"tool": "calculator", "params": {{"expression": "178.50 * 0.15"}}, "reason": "Calculate 15% of stock price"}}
{{"tool": "done", "params": {{}}, "reason": "All data gathered and message sent"}}
"""

    messages = [
        {"role": "system", "content": system_prompt},
        {
            "role": "user",
            "content": f"""Task: {task}

Current State:
{json.dumps(state, indent=2)}

What should I do next? Return only the JSON."""
        }
    ]

    data = {
        "model": "sonar-pro",
        "messages": messages,
        "temperature": 0.0  # Deterministic tool selection
    }

    for attempt in range(max_retries):
        try:
            print(f"Asking LLM for next action (attempt {attempt + 1}/{max_retries})...")

            response = requests.post(
                "https://api.perplexity.ai/chat/completions",
                headers=headers,
                json=data,
                timeout=30
            )
            response.raise_for_status()

            result = response.json()["choices"][0]["message"]["content"]
            print(f"LLM response received")

            # Clean and parse
            cleaned = clean_json_response(result)
            parsed = json.loads(cleaned)

            tool = parsed.get("tool", "").strip().lower()
            params = parsed.get("params", {})
            reason = parsed.get("reason", "")

            # Validate tool exists
            if tool in tool_registry.tools:
                print(f"Selected: {tool}")
                print(f"Reason: {reason}")
                if params:
                    print(f"Params: {params}")
                return {"tool": tool, "params": params, "reason": reason}
            else:
                print(f"Invalid tool '{tool}', retrying...")

        except json.JSONDecodeError as e:
            print(f"JSON parse error: {e}")
        except Exception as e:
            print(f"Error: {e}")

        if attempt < max_retries - 1:
            time.sleep(2 ** attempt)

    # Fallback
    print("All retries failed, returning 'done'")
    return {"tool": "done", "params": {}, "reason": "Fallback"}

# ========================================
# MULTI-TOOL AGENT
# ========================================

class MultiToolAgent:
    """
    Intelligent agent that can use multiple tools to complete complex tasks
    Uses REAL APIs for all data!
    """

    def __init__(self, task):
        self.task = task
        self.tool_registry = ToolRegistry()
        self.state = {
            "task": task,
            "steps_taken": [],
            "data_collected": {},
            "iteration": 0
        }
        self.max_iterations = 15

    def execute_action(self, action):
        """
        Execute an action (tool + parameters)

        Args:
            action (dict): Action with tool, params, reason

        Returns:
            Any: Result of tool execution
        """
        tool_name = action["tool"]
        params = action["params"]

        print(f"\n{'='*60}")
        print(f"EXECUTING: {tool_name}")
        print(f"{'='*60}")

        if tool_name == "done":
            return "complete"

        # Execute tool
        result = self.tool_registry.execute(tool_name, **params)

        # Store result in state
        self.state["data_collected"][tool_name] = result
        self.state["steps_taken"].append({
            "tool": tool_name,
            "params": params,
            "result": result
        })

        return result

    def run(self):
        """
        Main agent loop
        """
        print("="*60)
        print("MULTI-TOOL AGENT WITH REAL APIs")
        print("="*60)
        print(f"Task: {self.task}")
        print(f"Available Tools: {len(self.tool_registry.tools)}")
        print("="*60)

        for iteration in range(self.max_iterations):
            self.state["iteration"] = iteration + 1

            print(f"\n{'─'*60}")
            print(f"ITERATION {iteration + 1}/{self.max_iterations}")
            print(f"{'─'*60}")

            # Ask LLM for next action
            action = ask_llm_for_next_action(
                self.task,
                self.state,
                self.tool_registry
            )

            # Execute action
            result = self.execute_action(action)

            # Check if done
            if action["tool"] == "done" or result == "complete":
                print("\nTASK COMPLETE!")
                break

        # Final summary
        print(f"\n{'='*60}")
        print("FINAL SUMMARY")
        print(f"{'='*60}")
        print(f"Task: {self.task}")
        print(f"Tools used: {len(self.state['steps_taken'])}")
        print(f"Steps: {[s['tool'] for s in self.state['steps_taken']]}")
        print(f"{'='*60}")

# ========================================
# MAIN - TEST SCENARIOS
# ========================================

def main():
    print("\nDay 10: Multi-Tool Agent with REAL APIs\n")

    # Display API status
    print("API Configuration Status:")
    print(f"{'─'*60}")
    print(f"Perplexity: Active (LLM reasoning)")
    print(f"yfinance: Active (stock data - no key needed)")
    print(f"Google News RSS: Active (news - no key needed)")
    print(f"wttr.in: Active (weather fallback - no key needed)")
    print(f"Exchange Rate API: Active (currency - no key needed)")

    if NEWSAPI_KEY:
        print(f"NewsAPI: Active (better news quality)")
    else:
        print(f"NewsAPI: Not configured (using Google News)")

    if WEATHERAPI_KEY:
        print(f"WeatherAPI: Active (best weather data)")
    elif OPENWEATHER_KEY:
        print(f"OpenWeather: Active (good weather data)")
    else:
        print(f"Weather APIs: Not configured (using wttr.in)")

    if ALPHA_VANTAGE_KEY:
        print(f"Alpha Vantage: Active (financial data)")
    else:
        print(f"Alpha Vantage: Not configured (using yfinance only)")

    print(f"{'─'*60}\n")

    # TEST: Stock Analysis Task
    print("\n" + "*************************************")
    print("TEST: Stock Analysis with REAL Data")
    print("*************************************")

    agent = MultiToolAgent(
        task="tell me about some companies who were listed to nasdaq in past 6 months"
    )
    agent.run()


if __name__ == "__main__":
    main()
````

````python
"""
Day 11: Agent with Memory - FIXED VERSION
All API key loading and error handling issues resolved
"""

import os
import json
import time
import re
from typing import Dict, List, Any, Optional
from datetime import datetime
from collections import deque

# External libraries
import yfinance as yf
import requests

# ========================================
# API KEY LOADING - FIXED VERSION
# ========================================

print("="*60)
print("🔑 LOADING API KEYS")
print("="*60)

# Perplexity API (REQUIRED)
try:
    from google.colab import userdata
    PERPLEXITY_API_KEY = userdata.get("PERPLEXITY_API_KEY")
    if PERPLEXITY_API_KEY:
        print("✅ Perplexity API key loaded")
    else:
        print("❌ Perplexity API key is empty!")
        PERPLEXITY_API_KEY = None
except Exception as e:
    print(f"❌ Perplexity API key not found: {e}")
    PERPLEXITY_API_KEY = None

# WeatherAPI.com
try:
    from google.colab import userdata
    WEATHERAPI_KEY = userdata.get("WEATHERAPI_KEY")
    if WEATHERAPI_KEY:
        print("✅ WeatherAPI key loaded")
    else:
        WEATHERAPI_KEY = None
        print("⚠️  WeatherAPI key not found - using wttr.in fallback")
except:
    WEATHERAPI_KEY = None
    print("⚠️  WeatherAPI key not found - using wttr.in fallback")

# OpenWeather
try:
    from google.colab import userdata
    OPENWEATHER_KEY = userdata.get("OPENWEATHER_API_KEY")
    if OPENWEATHER_KEY:
        print("✅ OpenWeather key loaded")
    else:
        OPENWEATHER_KEY = None
        print("⚠️  OpenWeather key not found - using wttr.in fallback")
except:
    OPENWEATHER_KEY = None
    print("⚠️  OpenWeather key not found - using wttr.in fallback")

# Alpha Vantage
try:
    from google.colab import userdata
    ALPHA_VANTAGE_KEY = userdata.get("ALPHA_VANTAGE_KEY")
    if ALPHA_VANTAGE_KEY:
        print("✅ Alpha Vantage key loaded")
    else:
        ALPHA_VANTAGE_KEY = None
        print("⚠️  Alpha Vantage key not found - using yfinance only")
except:
    ALPHA_VANTAGE_KEY = None
    print("⚠️  Alpha Vantage key not found - using yfinance only")

# NewsAPI
try:
    from google.colab import userdata
    NEWSAPI_KEY = userdata.get("NEWSAPI_KEY")
    if NEWSAPI_KEY:
        print("✅ NewsAPI key loaded")
    else:
        NEWSAPI_KEY = None
        print("⚠️  NewsAPI key not found - using Google News RSS fallback")
except:
    NEWSAPI_KEY = None
    print("⚠️  NewsAPI key not found - using Google News RSS fallback")

print("="*60)
print()

# CRITICAL CHECK: Perplexity is required!
if not PERPLEXITY_API_KEY:
    print("❌❌❌ CRITICAL ERROR ❌❌❌")
    print("Perplexity API key is required but not found!")
    print()
    print("To fix this:")
    print("1. Go to: https://www.perplexity.ai/settings/api")
    print("2. Create an API key")
    print("3. In Colab: Click 🔑 (key icon) on left sidebar")
    print("4. Add secret: Name='PERPLEXITY_API_KEY', Value=<your key>")
    print()
    print("Agent will NOT work without this key!")
    print("="*60)
    # Don't exit - let user see the error message


# ========================================
# MEMORY COMPONENTS
# ========================================

class ConversationMemory:
    """Stores conversation history"""

    def __init__(self, max_history: int = 10):
        self.max_history = max_history
        self.history: List[Dict[str, str]] = []

    def add_user_message(self, message: str):
        self.history.append({
            "role": "user",
            "content": message,
            "timestamp": datetime.now().isoformat()
        })
        self._trim_history()

    def add_assistant_message(self, message: str):
        self.history.append({
            "role": "assistant",
            "content": message,
            "timestamp": datetime.now().isoformat()
        })
        self._trim_history()

    def _trim_history(self):
        if len(self.history) > self.max_history:
            self.history = self.history[-self.max_history:]

    def get_history(self) -> List[Dict[str, str]]:
        return self.history

    def get_history_string(self) -> str:
        if not self.history:
            return "No previous conversation."

        history_str = "Previous conversation:\n"
        for msg in self.history:
            role = "User" if msg["role"] == "user" else "Assistant"
            history_str += f"{role}: {msg['content']}\n"
        return history_str

    def clear(self):
        self.history = []


class EntityMemory:
    """Tracks entities mentioned in conversation"""

    def __init__(self):
        self.entities: Dict[str, Any] = {}

    def update_entity(self, entity_type: str, entity_id: str, data: Dict[str, Any]):
        if entity_type not in self.entities:
            self.entities[entity_type] = {}

        if entity_id not in self.entities[entity_type]:
            self.entities[entity_type][entity_id] = {
                "first_mentioned": datetime.now().isoformat(),
                "data": data
            }
        else:
            self.entities[entity_type][entity_id]["data"].update(data)

        self.entities[entity_type][entity_id]["last_updated"] = datetime.now().isoformat()

    def get_entity(self, entity_type: str, entity_id: str) -> Optional[Dict[str, Any]]:
        if entity_type in self.entities and entity_id in self.entities[entity_type]:
            return self.entities[entity_type][entity_id]["data"]
        return None

    def get_last_entity(self, entity_type: str) -> Optional[tuple]:
        entities = self.entities.get(entity_type, {})
        if not entities:
            return None

        latest = None
        latest_time = None

        for entity_id, entity_info in entities.items():
            updated = entity_info.get("last_updated")
            if not latest_time or updated > latest_time:
                latest_time = updated
                latest = (entity_id, entity_info["data"])

        return latest

    def get_summary(self) -> str:
        if not self.entities:
            return "No entities tracked yet."

        summary = "Tracked entities:\n"
        for entity_type, entities in self.entities.items():
            summary += f"\n{entity_type.upper()}:\n"
            for entity_id, entity_data in entities.items():
                summary += f"  - {entity_id}: {entity_data['data']}\n"
        return summary

    def clear(self):
        self.entities = {}


# ========================================
# MULTI-API WEATHER FUNCTIONS
# ========================================

def get_weather_wttr(city: str) -> Dict[str, Any]:
    """Weather from wttr.in (FREE, NO API KEY)"""
    try:
        url = f"https://wttr.in/{city}?format=j1"
        response = requests.get(url, timeout=5)
        response.raise_for_status()

        data = response.json()
        current = data['current_condition'][0]

        return {
            "city": city,
            "temperature": f"{current['temp_C']}°C",
            "temperature_f": f"{current['temp_F']}°F",
            "condition": current['weatherDesc'][0]['value'],
            "humidity": f"{current['humidity']}%",
            "source": "wttr.in"
        }
    except Exception as e:
        return {"error": str(e), "source": "wttr.in"}


def get_weather_weatherapi(city: str) -> Dict[str, Any]:
    """Weather from weatherapi.com"""
    if not WEATHERAPI_KEY:
        return {"error": "No API key configured", "source": "weatherapi.com"}

    try:
        url = "http://api.weatherapi.com/v1/current.json"
        params = {
            "key": WEATHERAPI_KEY,
            "q": city,
            "aqi": "no"
        }
        response = requests.get(url, params=params, timeout=5)
        response.raise_for_status()

        data = response.json()

        return {
            "city": data['location']['name'],
            "temperature": f"{data['current']['temp_c']}°C",
            "temperature_f": f"{data['current']['temp_f']}°F",
            "condition": data['current']['condition']['text'],
            "humidity": f"{data['current']['humidity']}%",
            "source": "weatherapi.com"
        }
    except Exception as e:
        return {"error": str(e), "source": "weatherapi.com"}


def get_weather_openweather(city: str) -> Dict[str, Any]:
    """Weather from openweathermap.org"""
    if not OPENWEATHER_KEY:
        return {"error": "No API key configured", "source": "openweathermap.org"}

    try:
        url = "http://api.openweathermap.org/data/2.5/weather"
        params = {
            "q": city,
            "appid": OPENWEATHER_KEY,
            "units": "metric"
        }
        response = requests.get(url, params=params, timeout=5)
        response.raise_for_status()

        data = response.json()

        temp_c = data['main']['temp']
        temp_f = (temp_c * 9/5) + 32

        return {
            "city": data['name'],
            "temperature": f"{temp_c}°C",
            "temperature_f": f"{temp_f:.1f}°F",
            "condition": data['weather'][0]['description'].title(),
            "humidity": f"{data['main']['humidity']}%",
            "source": "openweathermap.org"
        }
    except Exception as e:
        return {"error": str(e), "source": "openweathermap.org"}


def get_weather(city: str) -> Dict[str, Any]:
    """Multi-API weather with automatic fallback"""
    print(f"🌤️  Fetching weather for {city}...")

    # Try API 1: wttr.in (free, no key)
    print("   Trying wttr.in...")
    result = get_weather_wttr(city)
    if "error" not in result:
        print(f"   ✅ Success via {result['source']}")
        return result
    print(f"   ❌ Failed: {result['error'][:50]}...")

    # Try API 2: weatherapi.com
    if WEATHERAPI_KEY:
        print("   Trying weatherapi.com...")
        result = get_weather_weatherapi(city)
        if "error" not in result:
            print(f"   ✅ Success via {result['source']}")
            return result
        print(f"   ❌ Failed: {result['error'][:50]}...")

    # Try API 3: openweathermap.org
    if OPENWEATHER_KEY:
        print("   Trying openweathermap.org...")
        result = get_weather_openweather(city)
        if "error" not in result:
            print(f"   ✅ Success via {result['source']}")
            return result
        print(f"   ❌ Failed: {result['error'][:50]}...")

    # All APIs failed
    return {
        "error": "All weather APIs failed. Please try again later.",
        "city": city
    }


# ========================================
# OTHER TOOL FUNCTIONS
# ========================================

def get_stock_price(ticker: str) -> Dict[str, Any]:
    """Fetch current stock price"""
    print(f"📊 Fetching stock data for {ticker}...")
    try:
        stock = yf.Ticker(ticker)
        info = stock.info
        history = stock.history(period="5d")

        if history.empty:
            return {"error": f"No data available for {ticker}"}

        current_price = history['Close'].iloc[-1]
        if len(history) > 1:
            previous_close = history['Close'].iloc[-2]
            change_percent = ((current_price - previous_close) / previous_close) * 100
        else:
            change_percent = 0.0

        return {
            "ticker": ticker.upper(),
            "price": round(float(current_price), 2),
            "change_percent": round(float(change_percent), 2),
            "company_name": info.get('longName') or info.get('shortName') or ticker.upper(),
            "last_updated": str(history.index[-1].date())
        }
    except Exception as e:
        return {"error": str(e), "ticker": ticker.upper()}


def get_stock_history(ticker: str, years: int = 10) -> Dict[str, Any]:
    """Fetch historical stock price change"""
    print(f"📈 Fetching {years}-year history for {ticker}...")
    try:
        stock = yf.Ticker(ticker)
        period = f"{years}y"
        history = stock.history(period=period)

        if history.empty:
            return {"error": f"No historical data for {ticker}"}

        first_price = history['Close'].iloc[0]
        last_price = history['Close'].iloc[-1]
        change_amount = last_price - first_price
        change_percent = (change_amount / first_price) * 100

        return {
            "ticker": ticker.upper(),
            "period": f"{years} years",
            "first_price": round(float(first_price), 2),
            "first_date": str(history.index[0].date()),
            "current_price": round(float(last_price), 2),
            "current_date": str(history.index[-1].date()),
            "change_amount": round(float(change_amount), 2),
            "change_percent": round(float(change_percent), 2)
        }
    except Exception as e:
        return {"error": str(e), "ticker": ticker.upper()}


def calculator(expression: str) -> Dict[str, Any]:
    """Calculate mathematical expression"""
    print(f"🔢 Calculating: {expression}")
    try:
        allowed_chars = set("0123456789+-*/(). ")
        if not all(c in allowed_chars for c in expression):
            return {"error": "Invalid characters in expression"}

        result = eval(expression)
        return {
            "expression": expression,
            "result": float(result)
        }
    except Exception as e:
        return {"error": str(e), "expression": expression}


def send_message(message: str) -> Dict[str, bool]:
    """Send message to user"""
    print("\n" + "="*60)
    print("📨 MESSAGE TO USER:")
    print("="*60)
    print(message)
    print("="*60 + "\n")
    return {"success": True, "message": message}


# ========================================
# TOOL REGISTRY
# ========================================

class ToolRegistry:
    """Central registry of all available tools"""

    def __init__(self):
        self.tools = {
            "get_stock_price": {
                "function": get_stock_price,
                "description": "Get current stock price for a ticker symbol",
                "parameters": ["ticker"]
            },
            "get_stock_history": {
                "function": get_stock_history,
                "description": "Get historical stock price change over years",
                "parameters": ["ticker", "years"]
            },
            "get_weather": {
                "function": get_weather,
                "description": "Get current weather for a city",
                "parameters": ["city"]
            },
            "calculator": {
                "function": calculator,
                "description": "Perform mathematical calculations",
                "parameters": ["expression"]
            },
            "send_message": {
                "function": send_message,
                "description": "Send a detailed message to the user",
                "parameters": ["message"]
            },
            "done": {
                "function": None,
                "description": "Mark task as complete",
                "parameters": []
            }
        }

    def execute(self, tool_name: str, **kwargs) -> Any:
        if tool_name not in self.tools:
            return {"error": f"Unknown tool: {tool_name}"}

        tool = self.tools[tool_name]
        if tool["function"] is None:
            return {"status": "complete"}

        return tool["function"](**kwargs)

    def get_tool_list_for_llm(self) -> str:
        tool_list = "Available tools:\n"
        for name, tool in self.tools.items():
            params = ", ".join(tool["parameters"]) if tool["parameters"] else "none"
            tool_list += f"- {name}({params}): {tool['description']}\n"
        return tool_list


# ========================================
# AGENT WITH MEMORY
# ========================================

class AgentWithMemory:
    """Production-grade agent with full error handling"""

    def __init__(self, max_iterations: int = 15, max_history: int = 10):
        self.tool_registry = ToolRegistry()
        self.conversation_memory = ConversationMemory(max_history=max_history)
        self.entity_memory = EntityMemory()
        self.max_iterations = max_iterations
        self.api_call_times = deque(maxlen=5)

    def process_query(self, query: str) -> str:
        """Main entry point"""

        # CRITICAL: Check if API key is available
        if not PERPLEXITY_API_KEY:
            return (
                "❌ ERROR: Perplexity API key not configured!\n\n"
                "To fix:\n"
                "1. Go to: https://www.perplexity.ai/settings/api\n"
                "2. Create API key\n"
                "3. In Colab: Click 🔑 icon → Add secret\n"
                "4. Name: PERPLEXITY_API_KEY\n"
                "5. Value: <your key>"
            )

        print(f"\n{'='*60}")
        print(f"👤 USER QUERY: {query}")
        print(f"{'='*60}\n")

        self.conversation_memory.add_user_message(query)
        context = self._build_context(query)
        action = self._ask_llm_with_retry(query, context, max_retries=3)

        if not action:
            print("⚠️  LLM failed, trying fallback parser...")
            action = self._fallback_parser(query)

        if not action:
            response = (
                "I'm having trouble understanding that request. "
                "Please try rephrasing."
            )
            self.conversation_memory.add_assistant_message(response)
            return response

        result = self._execute_action(action)
        self._update_entity_memory(action, result)
        response = self._generate_response(action, result)
        self.conversation_memory.add_assistant_message(response)

        return response

    def _build_context(self, query: str) -> str:
        """Build context with memory"""
        context = ""

        if self.conversation_memory.get_history():
            context += self.conversation_memory.get_history_string() + "\n\n"

        if self.entity_memory.entities:
            context += self.entity_memory.get_summary() + "\n\n"

        context += f"Current query: {query}\n"
        return context

    def _wait_for_rate_limit(self):
        """Rate limiting"""
        if len(self.api_call_times) >= 5:
            time_since_5th = time.time() - self.api_call_times[0]
            if time_since_5th < 60:
                wait_time = 60 - time_since_5th + 2
                print(f"⏳ Rate limit: Waiting {wait_time:.0f}s...")
                time.sleep(wait_time)

        self.api_call_times.append(time.time())

    def _ask_llm_with_retry(self, query: str, context: str, max_retries: int = 3) -> Optional[Dict[str, Any]]:
        """Retry logic"""
        for attempt in range(max_retries):
            try:
                result = self._ask_llm_with_memory(query, context)
                if result:
                    return result

                if attempt < max_retries - 1:
                    wait_time = (attempt + 1) * 10
                    print(f"⚠️  Retry {attempt + 1}/{max_retries} in {wait_time}s...")
                    time.sleep(wait_time)
            except Exception as e:
                if attempt < max_retries - 1:
                    print(f"⚠️  Error: {e}")
                    time.sleep((attempt + 1) * 10)

        return None

    def _ask_llm_with_memory(self, query: str, context: str) -> Optional[Dict[str, Any]]:
        """Ask LLM with error handling"""

        self._wait_for_rate_limit()

        system_prompt = f"""You are an agent with memory. Respond ONLY in JSON format.

{self.tool_registry.get_tool_list_for_llm()}

Examples:
{{"tool": "get_weather", "params": {{"city": "NYC"}}, "reason": "Weather query"}}
{{"tool": "get_stock_price", "params": {{"ticker": "AAPL"}}, "reason": "Stock query"}}
{{"tool": "calculator", "params": {{"expression": "100*250"}}, "reason": "Math"}}

RESPOND WITH JSON ONLY."""

        user_message = f"{context}\n\nWhat should I do? JSON only."

        try:
            response = requests.post(
                "https://api.perplexity.ai/chat/completions",
                headers={
                    "Authorization": f"Bearer {PERPLEXITY_API_KEY}",
                    "Content-Type": "application/json"
                },
                json={
                    "model": "sonar-pro",
                    "messages": [
                        {"role": "system", "content": system_prompt},
                        {"role": "user", "content": user_message}
                    ],
                    "temperature": 0.0
                },
                timeout=30
            )

            if response.status_code == 429:
                print("❌ RATE LIMIT (429)")
                time.sleep(60)
                return None

            if response.status_code == 401:
                print("❌ API KEY ERROR (401) - Check your Perplexity API key!")
                return None

            if response.status_code != 200:
                print(f"❌ API Error ({response.status_code})")
                return None

            if not response.text:
                print("❌ Empty response")
                return None

            data = response.json()

            if "choices" not in data or not data["choices"]:
                print("❌ Invalid response structure")
                return None

            content = data["choices"][0]["message"]["content"]
            content = content.strip()

            # Clean JSON
            if content.startswith("```json"):
                content = content[7:]
            if content.startswith("```"):
                content = content[3:]
            if content.endswith("```"):
                content = content[:-3]
            content = content.strip()

            action = json.loads(content)

            if not isinstance(action, dict) or "tool" not in action:
                print(f"❌ Invalid action: {action}")
                return None

            print(f"✅ LLM: {action['tool']}({action.get('params', {})})\n")
            return action

        except json.JSONDecodeError as e:
            print(f"❌ JSON error: {e}")
            return None
        except Exception as e:
            print(f"❌ Error: {e}")
            return None

    def _fallback_parser(self, query: str) -> Optional[Dict[str, Any]]:
        """Fallback parser"""
        query_lower = query.lower()

        # Weather queries
        if any(word in query_lower for word in ['weather', 'temperature', 'climate']):
            # Extract city
            city_match = re.search(r'\b([A-Z][a-z]+(?:\s+[A-Z][a-z]+)*)\b', query)
            if city_match:
                return {
                    "tool": "get_weather",
                    "params": {"city": city_match.group(1)},
                    "reason": "Fallback: Weather query"
                }

            # Check if query has lowercase city name
            # Common patterns: "weather today", "weather in brooklyn"
            if "in" in query_lower:
                parts = query_lower.split("in")
                if len(parts) > 1:
                    city = parts[1].strip().split()[0].title()
                    return {
                        "tool": "get_weather",
                        "params": {"city": city},
                        "reason": "Fallback: Extracted city after 'in'"
                    }

            # Default to checking last weather entity
            last_weather = self.entity_memory.get_last_entity("weather")
            if last_weather:
                city, _ = last_weather
                return {
                    "tool": "get_weather",
                    "params": {"city": city},
                    "reason": "Fallback: Last weather city"
                }

            # If all else fails, suggest help
            return {
                "tool": "send_message",
                "params": {
                    "message": "Please specify a city. Example: 'weather in Brooklyn'"
                },
                "reason": "Fallback: Need city name"
            }

        # Stock queries
        if any(word in query_lower for word in ['stock', 'price']):
            ticker_match = re.search(r'\b([A-Z]{2,5})\b', query)
            if ticker_match:
                return {
                    "tool": "get_stock_price",
                    "params": {"ticker": ticker_match.group(1)},
                    "reason": "Fallback: Stock query"
                }

            last_stock = self.entity_memory.get_last_entity("stock")
            if last_stock:
                ticker, _ = last_stock
                return {
                    "tool": "get_stock_price",
                    "params": {"ticker": ticker},
                    "reason": "Fallback: Last stock"
                }

        # Historical queries
        if any(phrase in query_lower for phrase in ['change since', 'years', 'history']):
            years_match = re.search(r'(\d+)\s*years?', query_lower)
            years = int(years_match.group(1)) if years_match else 10

            last_stock = self.entity_memory.get_last_entity("stock")
            if last_stock:
                ticker, _ = last_stock
                return {
                    "tool": "get_stock_history",
                    "params": {"ticker": ticker, "years": years},
                    "reason": "Fallback: Historical query"
                }

        # Calculator
        if any(char in query for char in ['+', '-', '*', '/']):
            expr_match = re.search(r'[\d\.\+\-\*/\(\)\s]+', query)
            if expr_match:
                return {
                    "tool": "calculator",
                    "params": {"expression": expr_match.group(0).strip()},
                    "reason": "Fallback: Math"
                }

        print("⚠️  Fallback: No pattern matched")
        return None

    def _execute_action(self, action: Dict[str, Any]) -> Any:
        """Execute action"""
        tool_name = action.get("tool")
        params = action.get("params", {})

        if tool_name == "done":
            return {"status": "complete"}

        result = self.tool_registry.execute(tool_name, **params)
        print(f"🔧 Result: {result}\n")
        return result

    def _update_entity_memory(self, action: Dict[str, Any], result: Any):
        """Update entity memory"""
        tool_name = action.get("tool")
        params = action.get("params", {})

        if not isinstance(result, dict) or "error" in result:
            return

        if tool_name == "get_stock_price" and "ticker" in params:
            self.entity_memory.update_entity("stock", params["ticker"], result)
        elif tool_name == "get_stock_history" and "ticker" in params:
            self.entity_memory.update_entity("stock_history", params["ticker"], result)
        elif tool_name == "get_weather" and "city" in params:
            self.entity_memory.update_entity("weather", params["city"], result)

    def _generate_response(self, action: Dict[str, Any], result: Any) -> str:
        """Generate response"""
        tool_name = action.get("tool")

        if tool_name == "send_message":
            if isinstance(result, dict) and "message" in result:
                return result["message"]
            return str(result)

        if isinstance(result, dict) and "error" in result:
            return f"Sorry, error: {result['error']}"

        if tool_name == "get_stock_price":
            return (
                f"{result['company_name']} ({result['ticker']}) is at "
                f"${result['price']} ({result['change_percent']:+.2f}%)"
            )

        elif tool_name == "get_stock_history":
            return (
                f"{result['ticker']}: ${result['first_price']} ({result['first_date']}) → "
                f"${result['current_price']} ({result['current_date']}) = "
                f"{result['change_percent']:+.2f}% over {result['period']}"
            )

        elif tool_name == "get_weather":
            return (
                f"Weather in {result['city']}: {result['temperature']} / "
                f"{result['temperature_f']}, {result['condition']}, "
                f"Humidity: {result['humidity']} (via {result.get('source', 'weather API')})"
            )

        elif tool_name == "calculator":
            return f"{result['expression']} = {result['result']}"

        return str(result)

    def clear_memory(self):
        self.conversation_memory.clear()
        self.entity_memory.clear()
        print("🗑️  Memory cleared!")

    def show_memory(self):
        print("\n" + "="*60)
        print("🧠 MEMORY STATE")
        print("="*60)
        print("\n📝 CONVERSATION:")
        print(self.conversation_memory.get_history_string())
        print("\n💾 ENTITIES:")
        print(self.entity_memory.get_summary())
        print("="*60 + "\n")


# ========================================
# MAIN
# ========================================

def main():
    """Interactive interface"""
    print("="*60)
    print("🤖 DAY 11: AGENT WITH MEMORY (FIXED)")
    print("="*60)
    print()

    # Show API status
    print("📡 API Status:")
    print(f"  Perplexity: {'✅' if PERPLEXITY_API_KEY else '❌ REQUIRED!'}")
    print(f"  WeatherAPI: {'✅' if WEATHERAPI_KEY else '⚠️  Optional'}")
    print(f"  OpenWeather: {'✅' if OPENWEATHER_KEY else '⚠️  Optional'}")
    print()

    if not PERPLEXITY_API_KEY:
        print("❌ Cannot start: Perplexity API key required!")
        return

    agent = AgentWithMemory(max_history=10)

    print("Commands:")
    print("  Type query | 'memory' | 'clear' | 'quit'")
    print()

    while True:
        try:
            user_input = input("You: ").strip()

            if not user_input:
                continue

            if user_input.lower() == 'quit':
                print("👋 Goodbye!")
                break

            if user_input.lower() == 'memory':
                agent.show_memory()
                continue

            if user_input.lower() == 'clear':
                agent.clear_memory()
                continue

            response = agent.process_query(user_input)
            print(f"\n🤖 Agent: {response}\n")

        except KeyboardInterrupt:
            print("\n\n👋 Goodbye!")
            break
        except Exception as e:
            print(f"\n❌ Error: {e}\n")


if __name__ == "__main__":
    main()
````
